```python
import pandas as pd

df = pd.read_csv("house-price.csv")
# df[df['Address'].isnull()]
df = df.dropna(subset=['Address'])
print(df.isnull().sum())   # باید Address حالا 0 نشون بده
print(len(df))             # تعداد ردیف‌های باقی‌مونده
```

```python
df['Area'] = pd.to_numeric(df['Area'], errors='coerce')
```

```python
df[df['Area'] > 196.5]
```

```python
before = len(df)
df = df[df['Area'] <= 196.5]
after = len(df)

print(f"Count before removing: {before}")
print(f"Count after removing: {after}")
print(f"Removed: {before - after} Rowes ({(before-after)/before*100:.2f}%)")
```

```python
outlier_indices = df[df['Area'] > 196.5].index
df = df.drop(outlier_indices)
```
